# PREGUNTA 1

```python
import pandas as pd
from pathlib import Path
import gzip
```

```python
# -----------------------------------
# Cargar las rutas de manera relativa
# -----------------------------------
DATA_DIR = Path('../data')
DATA_FILE = DATA_DIR / 'data.csv.gz'

# ---------------------------------
# Validar la existencia del archivo
# ---------------------------------
assert DATA_FILE.exists()
```

```python
# ---------------------------------
# Identificar los separadores
# ---------------------------------
with gzip.open(DATA_FILE, "rt", encoding="utf-8") as f:
    lineas = f.read().splitlines()

print(repr(lineas[0]))
print(repr(lineas[1]))
```

```python
# ---------------------------------
# Leer los datos
# ---------------------------------
data = pd.read_csv(DATA_FILE, sep="\t", header=None)
data
```

```python
# -----------------------------------
# Suma Columna 1
# -----------------------------------
suma_columna1 = data[1].sum()
suma_columna1
```

# PREGUNTA 2

```python
conteo_letras = {}

for letra in data[0]:
    if letra in conteo_letras:
        conteo_letras[letra] += 1
    else:
        conteo_letras[letra] = 1
        print(conteo_letras)


print(letra, conteo_letras)

resultado = sorted(conteo_letras.items())
print(resultado)
```

# PREGUNTA 3

```python
dic_conteo = {}

for letra in data[0]:
    if letra not in dic_conteo:
        dic_conteo[letra] = data.loc[data[0] == letra, 1].sum()

resultado = sorted(dic_conteo.items())
print(resultado)
```

# PREGUNTA 4

```python
data = pd.read_csv(DATA_FILE, sep="\t", header=None)

data[2] = data[2].astype(str)
data["mes"] = data[2].str[5:7]
data["año"] = data[2].str[0:4]
data["dia"] = data[2].str[8:10]
```

```python
dict_mes = {}

for element in data["mes"]:
    if element in dict_mes:
        dict_mes[element] += 1
    else:
        dict_mes[element] = 1

resultado = sorted(dict_mes.items())
print(resultado)
```
