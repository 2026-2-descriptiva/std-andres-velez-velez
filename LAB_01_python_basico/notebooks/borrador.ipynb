# PREGUNTA 4

```python
data = pd.read_csv(DATA_FILE, sep="\t", header=None)

data[2] = data[2].astype(str)
data["mes"] = data[2].str[5:7]
data["año"] = data[2].str[0:4]
data["dia"] = data[2].str[8:10]
```

```python
dict_mes = {}

for element in data["mes"]:
    if element in dict_mes:
        dict_mes[element] += 1
    else:
        dict_mes[element] = 1

resultado = sorted(dict_mes.items())
print(resultado)
```

# PREGUNTA 5

```python
dict_max_min = {}

for letter in data[0]: 
   dict_max_min[letter] = (letter,
                           data.loc[data[0] == letter, 1].max(), 
                           data.loc[data[0] == letter, 1].min()
                           )


resultado = sorted(dict_max_min.values())
print(resultado)
```

# PREGUNTA 6

```python
# ----------------------------------------------------------------------------------
# Se crea metrics independeintes separando cada par "aaa: 1" usando , como separador
# ----------------------------------------------------------------------------------
data["metrics_independientes"] = data[4].str.split(",")
data
```

```python
# --------------------------------------------------------------
# Se separa la información a procesar en columnas independientes
# --------------------------------------------------------------
data_ampliada = data.explode("metrics_independientes")
data_ampliada["metrics_independientes_2"] = data_ampliada["metrics_independientes"].str.split(":")

data_ampliada["metrics_elemento_0"] = data_ampliada["metrics_independientes_2"].str[0]
data_ampliada["metrics_elemento_1"] = data_ampliada["metrics_independientes_2"].str[1]
data_ampliada["metrics_elemento_1"] = data_ampliada["metrics_elemento_1"].astype(int)
data_ampliada
```

```python
# Agrupamos por la métrica, calculamos min y max, y ordenamos el resultado final
resultado = list(map(tuple,
                    data_ampliada.
                    groupby("metrics_elemento_0")["metrics_elemento_1"].
                    agg(["min", "max"]).
                    reset_index().values
                    )
                )

print(resultado)
```

# PREGUNTA 10

```python
n_codes   = data[3].str.split(",").str.len() # Se crea una columa conformada por la longitud de las listas resultantes de separar data[3] por ,
n_metrics = data[4].str.split(",").str.len()

resultado = list(zip(data[0].tolist(), n_codes.tolist(), n_metrics.tolist()))   # Se usa zip apra agrupar las tres columnas: data[0], n_codes y n_metrics
print(resultado)
```

# PREGUNTA 11

```python
suma = (data.assign(letra = data[3].str.split(","))   # con data.assign se crea una copia de data que contiene una columna llamada letra, en donde los elmentos de data[3] son elementos de una lista
        .explode("letra")     # Generan nuevas filas en las que cada elemento de data[3] esta individualmente conservando la información de las demás
        .groupby("letra")[1]  # Se usa a letra como llave y a data[1] como valor
        .sum()
)

resultado = suma.to_dict()
resultado
```

# PREGUNTA 12

```python
resultado = (
    data.assign(llave_numero=data[4].str.split(","))   # Se usa assig para crear una copia de data en la cual se inserta una nueva columna (llave_numero). Esta columna es una lista que contiene los elementos resultantes de separar metrics por ,
        .explode("llave_numero")                       # Se usa explode para que cada elemento de llave_numero tenga so propia fila, respetando la información de las demás columnas 
        .assign(valor=lambda d: d["llave_numero"].str.split(":").str[1].astype(int))   # Se usa assign para crear una nueva columna llamada valor conformada por la parte numérica de llave_numero
        .groupby(0)["valor"]                           # groupby: se usa a data[0] como llave y data["valor"] como valor
        .sum()
        .to_dict()
)

resultado
```
