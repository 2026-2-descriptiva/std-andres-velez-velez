```python
import pandas as pd
from pathlib import Path
import sqlite3
```

```python
# DEFINICIÓN DE LAS RUTAS
DATA_DIR = Path.cwd().parent / "data"
DATA_TBL0 = DATA_DIR / "tbl0.csv.gz"
DATA_TBL1 = DATA_DIR / "tbl1.csv.gz"
DATA_TBL2 = DATA_DIR / "tbl2.csv.gz"

# DEFINICIÓN DE LOS NOMBRES DE LAS COLUMNAS
COLUMNS_TBL0 = ["K0", "c01", "c02", "c03", "c04"]
COLUMNS_TBL1 = ["K0", "K1", "c12", "c13", "c14", "c15", "c16"]
COLUMNS_TBL2 = ["K1", "c21", "c22", "c23", "c24", "c25"]

# LECTURA DE LA BASE DE DATOS
tbl0 = pd.read_csv(DATA_TBL0, compression="gzip", header=None, names=COLUMNS_TBL0)
tbl1 = pd.read_csv(DATA_TBL1, compression="gzip", header=None, names=COLUMNS_TBL1)
tbl2 = pd.read_csv(DATA_TBL2, compression="gzip", header=None, names=COLUMNS_TBL2)

# DEFINICIÓN DEL CONECTOR A LA BASE DE DATOS SQLITE3 (CONECTOR: ENTRE PYTHON Y SQLITE3)
conn = sqlite3.connect(":memory:")

# ESCRITURA DE LOS DATOS EN LA BASE DE DATOS SQLITE3
tbl0.to_sql("tbl0", conn, if_exists="replace", index=False)
tbl1.to_sql("tbl1", conn, if_exists="replace", index=False)
tbl2.to_sql("tbl2", conn, if_exists="replace", index=False)

tbl2
```

```python
# --------------------
# CONSULTA PREGUNTA 01
# --------------------

consulta_01 = """
SELECT SUM(c12) AS "SUM(c12)" 
FROM tbl1
"""
resultado_01 = pd.read_sql_query(consulta_01, conn)
resultado_01
```

```python
# --------------------
# CONSULTA PREGUNTA 02
# --------------------
consulta_02 = """
SELECT COUNT(K0) AS "COUNT(*)"
FROM tbl1
"""

resultado_02 = pd.read_sql_query(consulta_02, conn)
print(resultado_02)
```

```python
## --------------------
# CONSULTA PREGUNTA 03
# --------------------

consulta_03 = """
SELECT* 
FROM tbl1
ORDER BY c14 ASC
lIMIT 5
"""

resultado_03 = pd.read_sql_query(consulta_03, conn)
print(resultado_03)
```

```python
## --------------------
# CONSULTA PREGUNTA 04
# --------------------

consulta_04 = """
SELECT K0, c16
FROM tbl1
WHERE c16 LIKE K0 || '%'
"""
# WHERE c16 LIKE K0 || '%': hace las veces de expresión regular.
# WHERE c16 LIKE K0 || indica que el primer elemento de c16 debe ser igual al primer elemento de K0
# y el símbolo % indica que puede haber cualquier cantidad de caracteres después del primer elemento de c16.

resultado_04 = pd.read_sql_query(consulta_04, conn)
print(resultado_04)
```

```python
## --------------------
# CONSULTA PREGUNTA 05
# --------------------

consulta_05 = """
SELECT*
FROM tbl0
WHERE c02 = 100
OR c02 = 600
"""

resultado_05 = pd.read_sql_query(consulta_05, conn)
print(resultado_05)
```

```python
## --------------------
# CONSULTA PREGUNTA 12
# --------------------

consulta_12 = """
SELECT
    K0,
    MAX(c12) AS "MAX(c12)",
    MIN(c12) AS "MIN(c12)"
FROM tbl1
GROUP BY K0
"""

resultado_12 = pd.read_sql_query(consulta_12, conn)
print(resultado_12)
```

```python
## --------------------
# CONSULTA PREGUNTA 13
# --------------------

consulta_13 = """
SELECT
    K0,
    AVG(c12) AS "AVG(c12)"
FROM tbl1
WHERE c13 > 400
GROUP BY K0
"""

resultado_13 = pd.read_sql_query(consulta_13, conn)
print(resultado_13)
```

```python
## --------------------
# CONSULTA PREGUNTA 14
# --------------------

consulta_14 = """
SELECT
    tbl1.K0,
    AVG(tbl2.c21) AS "avg(c21)"
FROM tbl1
INNER JOIN tbl2
    ON tbl1.K1 = tbl2.K1
WHERE c13 > 400
GROUP BY tbl1.K0
"""

resultado_14 = pd.read_sql_query(consulta_14, conn)
print(resultado_14)
```
