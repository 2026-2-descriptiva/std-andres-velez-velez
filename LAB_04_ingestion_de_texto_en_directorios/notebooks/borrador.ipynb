```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR_TEST = Path.cwd().parent / "data" / "test"
DATA_DIR_TRAIN = Path.cwd().parent / "data" / "train"

SUBMISSION = Path.cwd().parent / "submission"
```

```python
registros_test = []

for carpeta in DATA_DIR_TEST.iterdir():
    if carpeta.is_dir():
        for archivo in carpeta.glob("*.txt"):
            frase = archivo.read_text(encoding="utf-8")
            registros_test.append({
                "phrase": frase,
                "target": carpeta.name
            })

df_test = pd.DataFrame(registros_test)
df_test["phrase,target"] = '"' + df_test["phrase"] + '"' + "," + df_test["target"]
df_test_final = df_test["phrase,target"]
df_test_final
```

```python
with open(SUBMISSION / "test_formato.csv", "w", encoding="utf-8") as f:
    f.write("phrase,target\n")
    f.write("\n".join(df_test_final))
```

```python
registros_train = []

for carpeta in DATA_DIR_TRAIN.iterdir():
    if carpeta.is_dir():
        for archivo in carpeta.glob("*.txt"):
            frase = archivo.read_text(encoding="utf-8")
            registros_train.append({
                "phrase": frase,
                "target": carpeta.name
            })

df_train = pd.DataFrame(registros_train)
df_train["phrase,target"] = '"' + df_train["phrase"] + '"' + "," + df_train["target"]
df_train_final = df_train["phrase,target"]
df_train_final
```

```python
with open(SUBMISSION / "train_formato.csv", "w", encoding="utf-8") as f:
    f.write("phrase,target\n")
    f.write("\n".join(df_train_final))
```
